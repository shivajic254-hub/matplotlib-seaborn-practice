```python
import matplotlib.pyplot as plt
import seaborn as sns 
%matplotlib inline 
```

```python
years = range(2000, 2006)
apples = [0.35, 0.6, 0.9, 0.8, 0.65, 0.8]
oranges = [0.4, 0.8, 0.9, 0.7, 0.6, 0.8]
```

```python
plt.plot(years,oranges)
```

```python
plt.bar(years,oranges)
```

```python
## we can do bar and line chart in the same line
plt.plot(years,oranges , 'o--r')
plt.bar(years,oranges)
plt.title('yield of oranges')
```

```python
## barplot with avarage
plt.bar(years , apples)
plt.bar(years , oranges , bottom = apples)
```

```python
## barplot with avarage bottom = apples need...
plt.bar(years , apples)
plt.bar(years , oranges )
```

```python
##^^  here the bottom = apples change/represents the whole apple representaion 
```

```python
## barplot with avarage
plt.bar(years , apples)
plt.bar(years , oranges , bottom = oranges)
```

```python
tips_df = sns.load_dataset('tips')
```

```python
tips_df
```

```python
bill_avg_df = tips_df.groupby('day')[['total_bill']].mean()
```

```python
bill_avg_df
```

```python
plt.bar(bill_avg_df.index ,bill_avg_df.total_bill);
```

```python
## bill_avg_df.index (x-axis): Uses the DataFrame's index as the category labels 
## along the horizontal axis—for example,
## 'Thur', 'Fri', 'Sat', and 'Sun' after .groupby('day')
## bill_avg_df.total_bill (y-axis): Uses the total_bill column values as the height of each bar
## along the vertical axis (such as the average bill per day).
```

```python
sns.barplot(x = 'day' ,y = 'total_bill' , hue ='sex' , data = tips_df);
```

```python
sns.barplot(y= 'day' , = 'total_bill' , hue ='sex' , data = tips_df);
```
