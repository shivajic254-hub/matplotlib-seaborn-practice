```python
import matplotlib.pyplot as plt
import seaborn as sns 
%matplotlib inline 
```

```python
flights = sns.load_dataset('flights')
```

```python
flights
```

```python
flights.describe()
```

```python
plt.plot(flights.year, flights.passengers)
```

```python
plt.plot(flights.passengers)
```

```python
flights = sns.load_dataset('flights').pivot(
    index='month', columns='year', values='passengers'
)
```

```python
flights
```

```python
# ^^ making this in ths matrix form 
```

```python
plt.title('no of passengers (1000s)')
sns.heatmap(flights)
```

```python
sns.heatmap(flights , fmt = 'd', annot = true)
```
